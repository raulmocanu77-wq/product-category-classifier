```python
df_clean = df.copy()
```

```python
df_clean.isnull().sum()
```

```python
(df_clean.isnull().mean() * 100).round(2)
```

```python
df_clean = df_clean.dropna(
    subset=["Product Title", "Category Label"]
)
```

```python
df_clean["Product Title"] = df_clean["Product Title"].astype(str).str.strip()

df_clean["Category Label"] = df_clean["Category Label"].astype(str).str.strip()

df_clean = df_clean[
    (df_clean["Product Title"] != "") &
    (df_clean["Category Label"] != "")
]
```

```python
print("Duplicate înainte:", df_clean.duplicated().sum())

df_clean = df_clean.drop_duplicates()

print("Duplicate după:", df_clean.duplicated().sum())
```

```python
df_clean["Product Title"] = (
    df_clean["Product Title"]
    .str.lower()
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)
```

```python
print("Dimensiune inițială:", df.shape)
print("Dimensiune după curățare:", df_clean.shape)

print("\nValori lipsă:")
print(df_clean.isnull().sum())

print("\nPrimele produse:")
display(df_clean.head())
```

```python
df_clean["Category Label"].value_counts()
```

```python
X = df_clean["Product Title"]
y = df_clean["Category Label"]

print("Număr de titluri:", len(X))
print("Număr de categorii:", y.nunique())

display(X.head())
display(y.head())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(
    max_features=10000,
    stop_words="english",
    ngram_range=(1, 2)
)

X_tfidf = tfidf.fit_transform(X)

print("Dimensiunea matricei TF-IDF:", X_tfidf.shape)
```

```python
print("Tipul datelor:", type(X_tfidf))
print("Număr de produse:", X_tfidf.shape[0])
print("Număr de caracteristici:", X_tfidf.shape[1])
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_tfidf,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
```

```python
print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)

print("Training labels:", y_train.shape)
print("Testing labels:", y_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression(
    max_iter=1000,
    random_state=42
)
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)

print("Primele 10 predicții:")
print(y_pred[:10])

print("\nCategoriile reale:")
print(y_test.iloc[:10].values)
```

```python
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)
```

```python
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", round(accuracy, 4))
print("Accuracy (%):", round(accuracy * 100, 2), "%")
```

```python
print(classification_report(y_test, y_pred))
```

```python
cm = confusion_matrix(y_test, y_pred)

print(cm)
```
